## Task 1: Semantic Similarity Comparison
### Objective:
Compare semantic similarity between pairs of sentences using cosine similarity and embeddings.

### Steps:
1. Load a pre-trained Sentence Transformer model.
2. Encode the sentence pairs.
3. Compute cosine similarity for each pair.

### Dataset:
- "A dog is playing in the park." vs. "A dog is running in a field."
- "I love pizza." vs. "I enjoy ice cream."
- "What is AI?" vs. "How does a computer learn?"


```python
%pip install sentence_transformers
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Sentence pairs
sentence_pairs = [
    ("A dog is playing in the park.", "A dog is running in a field."),
    ("I love pizza.", "I enjoy ice cream."),
    ("What is AI?", "How does a computer learn?")
]

# Compute similarities

for sentence1, sentence2 in sentence_pairs:

    # Convert both sentences into embeddings
    embeddings = model.encode([sentence1, sentence2])

    # Compare the two embeddings
    similarity = cosine_similarity(
        [embeddings[0]],
        [embeddings[1]])[0][0]

    print("Sentence 1:", sentence1)
    print("Sentence 2:", sentence2)
    print("Similarity:", round(similarity, 3))
    print()
```

The pizza and ice cream sentences had the highest similarity score at 0.528, although the dog sentences were very close at 0.522. The model appears to recognise the shared idea of enjoying food in the first pair and similar dog/activity/location concepts in the second pair. The AI and computer learning pair scored lower at 0.319, even though they are related topics.

Cosine similarity may miss meaning when sentences use negation, sarcasm or depend heavily on context. For example, “I love this” and “I do not love this” contain very similar words but have opposite meanings.

## Task 2: Document Clustering
### Objective:
Cluster a set of text documents into similar groups based on their embeddings.

### Steps:
1. Encode the documents using Sentence Transformers.
2. Use KMeans clustering to group the documents.
3. Analyze the clusters for semantic meaning.

```python
from sklearn.cluster import KMeans

# Documents to cluster
documents = [
    "What is the capital of France?",
    "How do I bake a chocolate cake?",
    "What is the distance between Earth and Mars?",
    "How do I change a flat tire on a car?",
    "What is the best way to learn Python?",
    "How do I fix a leaky faucet?"
]

# Convert all documents into embeddings
document_embeddings = model.encode(documents)

print(document_embeddings.shape)
```

```python
# Perform KMeans clustering

from sklearn.cluster import KMeans

# Create a clustering model with 3 groups
kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

# Fit the model to the document embeddings
clusters = kmeans.fit_predict(document_embeddings)

print(clusters)
```

```python
# Print cluster assignments

for document, cluster in zip(documents, clusters):
    print("Cluster", cluster, ":", document)
```

### Questions:
- How many clusters make the most sense? Why?
- Examine the documents in each cluster. Are they semantically meaningful? Can you assign a semantic "theme" to each cluster?
- Try this exercise with a larger dataset of your choice

Three clusters is a reasonable starting point because there are six documents and several broad topic types. However, the results are not perfectly grouped semantically. Cluster 1 is the clearest, containing two practical repair questions: changing a flat tire and fixing a leaky faucet. Cluster 0 mixes baking and space/distance, while Cluster 2 mixes geography and learning Python. This suggests that with such a small and varied dataset, KMeans does not find completely clean semantic themes.

## Task 3: Semantic Search System
### Objective:
Create a semantic search engine:
A user provides a query and you search the dataset for semantically relevant documents to return. Return the top 5 results.

### Dataset:
- Use the following set of documents:
    - "What is the capital of France?"
    - "How do I bake a chocolate cake?"
    - "What is the distance between Earth and Mars?"
    - "How do I change a flat tire on a car?"
    - "What is the best way to learn Python?"
    - "How do I fix a leaky faucet?"
    - "What are the best travel destinations in Europe?"
    - "How do I set up a local server?"
    - "What is quantum computing?"
    - "How do I build a mobile app?"


```python
import numpy as np

# Documents dataset
documents = [
    "What is the capital of France?",
    "How do I bake a chocolate cake?",
    "What is the distance between Earth and Mars?",
    "How do I change a flat tire on a car?",
    "What is the best way to learn Python?",
    "How do I fix a leaky faucet?",
    "What are the best travel destinations in Europe?",
    "How do I set up a local server?",
    "What is quantum computing?",
    "How do I build a mobile app?"
]

# Compute document embeddings

doc_embeddings = model.encode(documents)

print(doc_embeddings.shape)
```

```python
# Create the search function
#This function should encode the user query and return the top N documents that most resemble it
def semantic_search(query, documents, doc_embeddings, top_n=5):

    # Convert the user's query into an embedding
    query_embedding = model.encode([query])

    # Compare the query with every document
    similarities = cosine_similarity(
        query_embedding,
        doc_embeddings
    )[0]

    # Sort document positions from highest similarity to lowest
    ranked_indices = np.argsort(similarities)[::-1]

    # Keep only the top N results
    top_indices = ranked_indices[:top_n]

    # Print the results
    for index in top_indices:
        print("Document:", documents[index])
        print("Similarity:", round(similarities[index], 3))
        print()
```

```python
# Test the search function
query = "Explain programming languages."
semantic_search(query, documents, doc_embeddings)
```

### Questions:

- How can you improve the ranking explanation for users?
- Try this approach with a larger dataset

The search returns documents based on semantic similarity rather than exact keyword matching. The top result was “What is quantum computing?” with a similarity of 0.435, followed by “What is the best way to learn Python?” at 0.319. Some of the lower-ranked results are only weakly related, which shows that semantic search can still return imperfect matches when the dataset is small or the query is broad.


How can you improve the ranking explanation for users?
The results could be improved by showing the similarity score alongside each result and giving a short explanation of why the document was considered relevant. A minimum similarity threshold could also be used so that very weak matches are not shown.
